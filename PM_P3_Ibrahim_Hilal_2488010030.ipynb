## Latihan Mandiri
**Latihan 1.** Muat dataset lain bawaan seaborn, yaitu `tips` (`sns.load_dataset('tips')`). Tampilkan `head()`, `shape`, dan `info()`.


```python
# Latihan 1: Memuat dataset tips dan menampilkan head, shape, serta info
import seaborn as sns

tips = sns.load_dataset('tips')

print("--- 5 Baris Pertama (head) ---")
display(tips.head())

print("\n--- Ukuran Dataset (shape) ---")
print(tips.shape)

print("\n--- Informasi Struktur Data (info) ---")
tips.info()
```

**Latihan 2.** Pada dataset `tips`, buat histogram kolom `total_bill` dan box plot kolom `tip`. Amati sebaran dan outlier-nya.


```python
# Latihan 2: Histogram total_bill dan Box plot tip
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 5))

# Histogram total_bill
plt.subplot(1, 2, 1)
sns.histplot(tips['total_bill'], kde=True, color='skyblue')
plt.title('Histogram Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frekuensi')

# Box plot tip
plt.subplot(1, 2, 2)
sns.boxplot(y=tips['tip'], color='lightgreen')
plt.title('Box Plot Tip (Deteksi Outlier)')
plt.ylabel('Tip')

plt.tight_layout()
plt.show()
```

**Latihan 3.** Buat scatter plot `total_bill` (x) vs `tip` (y) pada dataset `tips`, diwarnai menurut kolom `time` (Lunch/Dinner). Apa yang dapat Anda simpulkan?


```python
# Latihan 3: Scatter plot total_bill vs tip diwarnai menurut time
import matplotlib.pyplot as plt
import seaborn as sns

# Memuat dataset tips
tips = sns.load_dataset('tips')

plt.figure(figsize=(8, 5))
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time')
plt.title('Scatter Plot Total Bill vs Tip (by Time)')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()

# Kesimpulan:
# Terdapat korelasi positif antara total_bill dan tip (semakin besar tagihan, semakin besar tip yang diberikan).
# Transaksi pada waktu Dinner lebih mendominasi dan cenderung menghasilkan total bill serta tip yang lebih tinggi dibandingkan Lunch.
```

**Latihan 4.** Buat heatmap korelasi untuk fitur numerik pada dataset `tips`. Pasangan fitur mana yang paling berkorelasi?


```python
# Latihan 4: Heatmap korelasi untuk fitur numerik pada dataset tips
import matplotlib.pyplot as plt
import seaborn as sns

# Memuat dataset tips
tips = sns.load_dataset('tips')

plt.figure(figsize=(6, 4))
sns.heatmap(tips.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Heatmap Korelasi Fitur Numerik Dataset Tips')
plt.show()

# Pasangan fitur yang paling berkorelasi:
# Fitur total_bill dan tip memiliki korelasi tertinggi (r ≈ 0.68).
```

### Jawaban Refleksi Praktikum Pertemuan 3

1. **Mengapa EDA penting dilakukan sebelum membangun model?**
   EDA penting untuk memahami karakteristik dasar data, memeriksa kualitas data (seperti menemukan nilai yang hilang atau *outlier*), memahami pola sebaran fitur, serta melihat hubungan antarvariabel. Hal ini membantu kita menentukan langkah praproses (*preprocessing*) yang tepat dan memilih algoritma model yang paling sesuai.

2. **Apa perbedaan informasi yang Anda peroleh dari histogram, box plot, dan scatter plot?**
   - **Histogram:** Menampilkan sebaran frekuensi dan bentuk distribusi dari satu fitur numerik (misalnya simetris atau miring).
   - **Box Plot:** Menampilkan nilai ringkasan statistik (kuartil, median) serta mendeteksi keberadaan *outlier* (data pencilan) pada fitur.
   - **Scatter Plot:** Menampilkan hubungan/korelasi antara dua fitur numerik serta pola pemisahan kelompok data berdasarkan kategori (*species*).

3. **Temuan EDA mana pada dataset Iris yang menurut Anda paling berguna untuk klasifikasi? Mengapa?**

Melalui *scatter plot* dan *pairplot*, kedua fitur ini menunjukkan pemisahan kelompok spesies yang sangat jelas (spesies *setosa* terpisah sempurna, sedangkan *versicolor* dan *virginica* memiliki batas pemisah yang tegas). Fitur ini memiliki korelasi tinggi terhadap pemisahan kelas sehingga paling informatif untuk digunakan oleh algoritma klasifikasi.
